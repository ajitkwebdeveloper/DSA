### Reverse Linked List ⭐ Amazon Favorite

input : 1 -> 2 -> 3 -> 4 <br>

output: 4 -> 3 -> 2 -> 1 <br>

solution: Need 3 pointers: <br>
    prev  <br>
    curr <br>
    next 

Reverse pointer one by one.

```python
class Node:
    """A class to represent a node in the linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    """A class to represent a singly linked list."""
    def __init__(self):
        self.head = None

    def insert_at_beginning(self, data):
        """Inserts a new node at the beginning of the linked list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        """Displays the linked list elements."""
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")



    """
    Reverse the linked list in place ⭐⭐⭐
    Time Complexity: O(n)
    Space Complexity: O(1)
    
    """
    def reverse_linked_list(self, head):
        
        """Reverses the linked list."""

        prev = None
        current = head

        while current:
            next_node = current.next  # Store the next node
            current.next = prev       # Reverse the link
            prev = current            # Move prev to current
            current = next_node       # Move to the next node

        return prev  # New head of the reversed list



"""
⭐⭐⭐
Find the middle node of the linked list using the two-pointer technique.
Time: O(n)
Space: O(1)

"""
def find_middle_node(head):

    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow


"""
Detect Cycle ⭐⭐⭐
Floyd Cycle Detection
Time: O(n)
1 -> 2 -> 3 -> 4
     ^       |
     |_______|

"""

def detect_cycle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            return True  # Cycle detected

    return False  # No cycle detected
```

```python
ll = LinkedList()

ll.insert_at_beginning(25)
ll.insert_at_beginning(20)
ll.insert_at_beginning(15)
ll.insert_at_beginning(10)
ll.insert_at_beginning(5)

ll.display()

print("---------------------------------")
middle_node = find_middle_node(ll.head)
print(f"middle node : {middle_node.data}")

print("---------------------------------")

print("Reversing the linked list...")
reversed_head = ll.reverse_linked_list(ll.head)
ll.head = reversed_head
ll.display()  # This displays the reversed list.

print("---------------------------------")
```
