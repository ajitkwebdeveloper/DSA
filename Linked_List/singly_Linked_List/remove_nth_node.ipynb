##### Given a linked list, remove the Nth node from the end and return the head.

input : 1 -> 2 -> 3 -> 4 -> 5 <br>
n = 2 <br>

The 2nd node from the end is 4. <br>

After deleting: <br>

1 -> 2 -> 3 -> 5

##### Optimal Approach : Two pointers
slow pointer <br>
fast pointer

Keep a gap of n nodes between them.When fast reaches the end:<br>
slow will be just before the node to delete

```python
class Node:
    """A class to represent a node in the linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    """A class to represent a singly linked list."""
    def __init__(self):
        self.head = None

    def insert_at_beginning(self, data):
        """Inserts a new node at the beginning of the linked list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        """Displays the linked list elements."""
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


    """
    Removes the nth node from the linked list.
    """

    def remove_nth_node(self, head, n):

        dummy = Node(0)
        dummy.next = head

        slow = dummy
        fast = dummy

        for _ in range(n + 1):
            # move fast pointer n+1 steps ahead
            # to keep gap of n between slow and fast
            fast = fast.next

        
        while fast:
            # move both pointers until fast reaches the end
            slow = slow.next
            fast = fast.next

        slow.next = slow.next.next
        
```

```python
ll = LinkedList()

ll.insert_at_beginning(5)
ll.insert_at_beginning(4)
ll.insert_at_beginning(3)
ll.insert_at_beginning(2)
ll.insert_at_beginning(1)
ll.display()

ll.remove_nth_node(ll.head, 2)  # Remove the 2nd node from the end
ll.display()  # Display the updated linked list
```
