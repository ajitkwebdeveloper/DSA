#### 
  Merge Two Sorted Lists ⭐⭐⭐ <br> <br>
  Input:  
  1 -> 3 -> 5 <br>
  2 -> 4 -> 6  <br>

   Output: <br>
    1 -> 2 -> 3 -> 4 -> 5 -> 6


```python
class Node:
    """A class to represent a node in the linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    """A class to represent a singly linked list."""
    def __init__(self):
        self.head = None

    def insert_at_beginning(self, data):
        """Inserts a new node at the beginning of the linked list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        """Displays the linked list elements."""
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


    def merge_sorted_lists(self, l1, l2):
        """Merges two sorted linked lists into one sorted linked list."""

        dummy = Node(0)
        tail = dummy

        while l1 and l2:
            if l1.data < l2.data:
                tail.next = l1
                l1 = l1.next
            else:
                tail.next = l2
                l2 = l2.next

            tail = tail.next

        tail.next = l1 or l2
        return dummy.next

 
```

```python
ll = LinkedList()

ll.insert_at_beginning(5)
ll.insert_at_beginning(3)
ll.insert_at_beginning(1)
ll.display()

l2 = LinkedList()
l2.insert_at_beginning(6)
l2.insert_at_beginning(4)
l2.insert_at_beginning(2)
l2.display()

print("---------------------------------")
print("Merging the two sorted linked lists...")
merged_head = ll.merge_sorted_lists(ll.head, l2.head)
merged_list = LinkedList()
merged_list.head = merged_head
merged_list.display()
```
