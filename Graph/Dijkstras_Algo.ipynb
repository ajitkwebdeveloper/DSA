##### Dijkstra is used to find the shortest path from a source node to all other nodes in a weighted graph.

When to Use

✅ Weighted Graph

✅ Positive Weights Only




#### PseudoCode

heap = [(0, source)]

while heap:

    current_distance, node = pop minimum

    for neighbor, weight:

        new_distance =
            current_distance + weight

        if new_distance < distance[neighbor]:

                new_distance

            push(new_distance, neighbor)

```python
import heapq
def dijkstra(graph, start):

    distances = {}
    for node in graph:
        distances[node] = float('inf') # Initialize distances to infinity

    distances[start] = 0 # Distance to the start node is 0

    heap = [(0, start)] # Min-heap to store (distance, node) pairs

    while heap:

        current_distance, current_node = heapq.heappop(heap)
        
        if current_distance > distances[current_node]:
            continue

        for neighbor, weight in graph[current_node]:

            new_distance = current_distance + weight

            if new_distance < distances[neighbor]: # Check if the new distance is shorter

                distances[neighbor] = new_distance
                heapq.heappush(heap, (new_distance, neighbor))
                print(heap)

    return distances




```

```python
graph = {
    "A": [("B",4), ("C",1), ("D",7)],
    "B": [("A",4), ("D",2)],
    "C": [("A",1), ("D",5)],
    "D": [("A",7), ("B",2), ("C",5)]
}

print(dijkstra(graph, "A"))
```
