Breadth First Search (BFS) in Graph

BFS (Breadth First Search) is a graph traversal algorithm that visits nodes level by level.

It uses a Queue (FIFO).

       A
      / \
     B   C
    / \   \
   D   E   F

output:   A → B → C → D → E → F

First visit A <br>
Then all neighbors of A (B, C)  <br>
Then neighbors of B (D, E)  <br>
Then neighbors of C (F) 

=============== BFS Pseudocode=================

1. Create Queue
2. Create Visited Set

3. Add start node to Queue
4. Mark start node as Visited

5. While Queue is not empty:

       Remove front node

       Process current node

       For each neighbor:
            If not visited:
                Mark visited
                Add to Queue

=================== Time Complexity ==================================
Vertices = V <br>
Edges = E

Each node is visited once.

Each edge is checked once.

Time Complexity = O(V + E)

```python
from collections import deque

def bfs(graph, start):

    queue = deque()
    visited = set()

    queue.append(start)
    visited.add(start)

    while queue:
        current_node = queue.popleft()
        print(current_node, end=" ")

        for neighbor in graph[current_node]:
            if neighbor not in visited:
                queue.append(neighbor)
                visited.add(neighbor)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

bfs(graph, "A")
```
