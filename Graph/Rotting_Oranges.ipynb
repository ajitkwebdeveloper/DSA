#### Rotting Oranges :    Multi-Source BFS

You are given a grid:<br>

0 = Empty Cell <br>
1 = Fresh Orange <br>
2 = Rotten Orange <br>

Every minute: <br>
 
A rotten orange infects all
adjacent fresh oranges <br>
(up, down, left, right)

Return: <br>
Minimum minutes needed to rot all oranges




#### ++++++++++ pseudocode ++++++++++++++++++++
queue = all rotten oranges

fresh = count fresh oranges

minutes = 0

while queue not empty
      and fresh > 0

    size = len(queue)

    process current level

    for each rotten orange

         rot neighbors

         add new rotten oranges

    minutes += 1

if fresh > 0
      return -1

return minutes


```python
# Time Complexity = O(m × n)
# space Complexity = O(m × n)

from collections import deque
import queue

def orangesRotting(grid):

    rows = len(grid)
    cols = len(grid[0])

    queue = deque()
    fresh = 0

     # Step 1:
    # Find rotten oranges
    # Count fresh oranges

    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 2:
                queue.append((r,c))

            elif grid[r][c] == 1:
                fresh += 1

    print(f"Initial fresh oranges: {fresh}")
    print(queue)

    # No fresh oranges
    if fresh == 0:
        return 0

    directions = [
                   (1,0),   # down
                   (-1,0),  # up
                   (0,1),   # right
                   (0,-1)   # left
                   ]
    minutes = 0
    while queue and fresh > 0:

        level_size = len(queue)

        for _ in range(level_size):
            r, c = queue.popleft()

            for dr, dc in directions:
                nr = r + dr
                nc = c + dc

                if(
                    0 <= nr < rows and
                    0 <= nc < cols and
                    grid[nr][nc] == 1
                ):
                    grid[nr][nc] = 2
                    fresh -= 1
                    queue.append((nr, nc))

            minutes += 1

    return minutes if fresh == 0 else -1


    

grid = [
    [2,1,1],
    [1,1,0],
    [0,1,1]
]

print(orangesRotting(grid))
```
