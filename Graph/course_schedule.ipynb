#### Course Schedule

#### problem: 
There are numCourses courses: 0, 1, 2, 3, ... <br>
You are given prerequisites: <br>
prerequisites = [[a, b]] <br>

Meaning: <br>
To take course a,
you must first complete course b

Edge: b → a

##### What Are We Really Checking?
We are checking: Does the directed graph contain a cycle?

If cycle exists:   Cannot finish courses


##### Algorithm Used
DFS Cycle Detection in Directed Graph

Time Complexity: O(V + E)

V = Courses
E = Prerequisites


```python
from collections import defaultdict
def canFinish(numCourses, prerequisites):

    graph = defaultdict(list)

    # build the graph
    for course, prereq in prerequisites:
        graph[course].append(prereq)
    print(f"Graph representation of courses and prerequisites: {dict(graph)}")


    visited = set()
    path = set()

    def dfs(course):

        # cycle found
        if course in path:
            return True

        # already visited
        if course in visited:
            return False

        path.add(course)

        for prereq in graph[course]:
            print(f"Visiting prerequisite {prereq} for course {course}")
            if dfs(prereq):
                return True

        print(f"Course {course} has no cycles, marking as visited.")   
        print(f"Current path: {path}")
   
        path.remove(course)
        visited.add(course)
        print(f"Current visited: {visited}") 

        return False


    for course in range(numCourses):
        print(f"Checking course: {course}")

        if dfs(course):
            return False

    return True

numCourses = 4
prerequisites = [
    [1,0],
    [2,1],
    [3,2]
]

print(canFinish(numCourses, prerequisites))
```
