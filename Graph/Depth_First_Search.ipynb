DFS (Depth First Search) is a graph traversal algorithm that explores a path as deep as possible before backtracking.

Go Deep → Go Deep → Go Deep
No more nodes?
Backtrack

        A
       / \
      B   C
     / \
    D   E

output: A → B → D → E → C



1. Visit current node

2. Mark node as visited

3. For each neighbor:

       If neighbor not visited:

            DFS(neighbor)

##### DFS Recursive

Time  = O(V + E) <br>
Space = O(V)

```python
def dfs(graph, node, visited):

    visited.add(node)
    print(node, end= " ")

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

visited = set()  # Set to keep track of visited nodes.
dfs(graph, "A", visited)
```
