#### Cycle Detection in Graph Using DFS
A cycle exists if you can start from a node and come back to the same node.

##### ++++++ DFS Algorithm +++++++++
For every node:

If not visited:

    Run DFS(node, parent)
DFS(node, parent):

Mark node visited

For each neighbor:

    If neighbor not visited:

        DFS(neighbor, node)

    Else if neighbor != parent:

        Cycle Found



```python
from collections import defaultdict

def has_cycle(n, edges):

  graph = defaultdict(list)
  for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)

  visited = set()

  def dfs(node, parent):

    visited.add(node)

    for neighbor in graph[node]:
      if neighbor not in visited:
        if dfs(neighbor, node):
          return True

        elif neighbor != parent:
          return True


    return False


  for node in range(n):
    if node not in visited:
      if dfs(node, -1):
        return True

  return False



n = 4

edges = [
    [0,1],
    [1,2],
    [2,3],
    [3,0]
]

print(has_cycle(n, edges))
```
