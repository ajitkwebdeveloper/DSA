#### Open The Lock
Problem Statement

You have a lock: 0000

Each wheel can:

0 → 1
1 → 2
...
8 → 9
9 → 0

or backward:
0 → 9
9 → 8
...
1 → 0
<br>
you are given: <br>
deadends = ["0201","0101","0102","1212","2002"]

target = "0202"

Find the minimum number of turns needed to reach the target.

#### ++++ Algorithm +++++++++++++++

Put "0000" into queue

visited = {"0000"}

moves = 0

While queue not empty:

    Process all nodes in current level

    If target found:
         return moves

    Generate 8 neighbors

    Ignore:
        visited states
        deadends

    Add valid neighbors

    moves += 1

If queue becomes empty:

    return -1

```python
from collections import deque
def openLock(deadends, target):
    dead = set(deadends)

    if '0000' in dead:
        return -1

    queue = deque(["0000"])
    visited = set("0000")
    moves = 0

    while queue:

        size = len(queue)

        for _ in range(size):
            state = queue.popleft()
           
            if state == target:
                return moves

            # Generate all possible next states by turning each wheel up or down
            for i in range(4):
                digit = int(state[i])

                # Turn the wheel up
                up = (digit + 1) % 10
                next_state = state[:i] + str(up) + state[i+1:]

                if next_state not in dead and next_state not in visited:
                    visited.add(next_state)
                    queue.append(next_state)
                    

                # Turn the wheel down
                down = (digit - 1) % 10
                next_state = state[:i] + str(down) + state[i+1:]

                if next_state not in dead and next_state not in visited:
                    visited.add(next_state)
                    queue.append(next_state)
              
        moves += 1

    return -1



```

```python
deadends = ["0201","0101","0102","1212","2002"]
target = "0202"

print(openLock(deadends, target))
```
