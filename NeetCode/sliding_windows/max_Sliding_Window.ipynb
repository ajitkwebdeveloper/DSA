### Sliding Window Maximum

##### Given an array nums and an integer k, return the maximum element in every window of size k.

[1, 3, -1]      max = 3 <br>
[3, -1, -3]     max = 3   <br>
[-1, -3, 5]     max = 5  <br>
[-3, 5, 3]      max = 5  <br>
[5, 3, 6]       max = 6  <br>
[3, 6, 7]       max = 7  <br>

output: [3, 3, 5, 5, 6, 7]

##### Brute force approach

Time  = O(n*k)
Space = O(1)

```python
def maxSlidingWindow(nums, k):
    result = []

    for i in range(len(nums) - k + 1):
        result.append(max(nums[i:i+k]))

    return result

nums = [1,3,-1,-3,5,3,6,7]
k = 3
print(maxSlidingWindow(nums, k))
```
