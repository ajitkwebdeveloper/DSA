#### Given two strings s and t, return true if the two strings are anagrams of each other, otherwise return false.

Input: s = "racecar", t = "carrace"

Output: true

```python
# 1. Sorting

def isAnagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    return sorted(s) == sorted(t)

s = "listen"
t = "silent"

print(isAnagram(s, t)) 
```

```python
# 2. Hash Map

def isAnagram(s: str, t: str) -> bool:

    if len(s) != len(t):
        return False

    count = {}

    for char in s:
        count[char] = count.get(char, 0) + 1

    for char in t:
        if char not in count:
            return False
        count[char] -= 1

        if count[char] < 0:
            return False
        
    return True


s = "listen"
t = "silent"

print(isAnagram(s, t)) 
```
